```python
import importlib

missing = []
for name in ("numpy", "sympy", "matplotlib", "torch"):
    try:
        mod = importlib.import_module(name)
        print(f"{name:11} {getattr(mod, '__version__', '?')}")
    except ImportError:
        missing.append(name)
print("missing:", ", ".join(missing) if missing else "nothing")
```
